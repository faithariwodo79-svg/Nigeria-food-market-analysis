### Data Quality Issue: Unmatched Market ID

One market ID (`1037`) appears in the food-price dataset but is not present in the market dimension table.

A total of **53 observations** are associated with this unmatched market ID.

Because these records cannot be reliably linked to a market name or geographic information, they were excluded from the primary analytical dataset.

This ensures that the final dataset can be reliably joined with the available market metadata.

```python
orphan_records = food_prices[~food_prices["market_id"].isin(markets["market_id"])].copy()

"""
Saving the 53 observations associated with market ID 1037 lacking a corresponding
market_dimension record
"""
```

```python
food_prices = food_prices[food_prices["market_id"].isin(markets["market_id"])].copy()
```

### Standardizing Price Units

Some observations are recorded on a gram-based unit.

To make prices comparable across observations, gram-based prices were converted to an equivalent price per kilogram by multiplying the normalized price by **1,000**.

This standardization allows the KG-based observations to be analyzed on a consistent unit basis.

```python
# Converting grams to kilograms so normalized_price is comparable
mask_grams = food_prices["normalized_unit"] == "G"

food_prices.loc[mask_grams, "normalized_price"] = food_prices.loc[mask_grams, "normalized_price"] * 1000
food_prices.loc[mask_grams, "normalized_unit"] = "KG"
```

```python
print(food_prices["normalized_unit"].value_counts())
```

```python
print("Non-positive raw prices:",(food_prices["price"] <= 0).sum())
print("Non-positive normalised prices:",(food_prices["normalized_price"] <= 0).sum())
```

```python
# Primary analysis dataset - KG based commodities only
df_main = food_prices[food_prices["normalized_unit"] == "KG"].copy()

# Optional secondary dataset - liquids
df_liquids = food_prices[food_prices["normalized_unit"] == "L"].copy()
```
